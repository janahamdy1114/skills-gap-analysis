```python
import pandas as pd

df = pd.read_csv("../data/raw/ml_jobs.csv")

print("Dataset shape:", df.shape)

display(df.head())
```

```python
print("Columns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Duplicate descriptions
print(
    "Duplicate descriptions:",
    df["job_description_text"].duplicated().sum()
)

# Missing descriptions
print(
    "Missing descriptions:",
    df["job_description_text"].isna().sum()
)

# Seniority distribution
print(df["seniority_level"].value_counts(dropna=False))
```

```python
for i in range(min(3, len(df))):
    print(f"\n{'='*60}")
    print("Job title:", df.iloc[i]["job_title"])
    print("Company:", df.iloc[i]["company_name"])
    print("Seniority:", df.iloc[i]["seniority_level"])
    print("\nDescription:")
    print(df.iloc[i]["job_description_text"])
```
