```python
import pandas as pd
import re

# Load the final 30-job dataset
df = pd.read_csv("../data/processed/selected_30_jobs.csv")

# Assign IDs consistent with your ingestion script
df["job_id"] = [f"job_{i:03d}" for i in range(1, len(df) + 1)]


def find_keyword_matches(description, keywords):
    """Return the keywords found in a job description."""
    description = str(description)
    matched = []

    for keyword in keywords:
        # Special handling for programming-language names
        if keyword.lower() == "c":
            pattern = r"(?<!\w)C(?!\w)"
        elif keyword.lower() == "c++":
            pattern = r"(?<!\w)C\+\+(?!\w)"
        else:
            pattern = r"(?<!\w)" + re.escape(keyword) + r"(?!\w)"

        if re.search(pattern, description, flags=re.IGNORECASE):
            matched.append(keyword)

    return matched


def suggest_relevant_jobs(query, keywords, dataset=df):
    """Rank jobs according to how many query keywords they contain."""
    results = []

    for _, row in dataset.iterrows():
        matches = find_keyword_matches(
            row["cleaned_description"], keywords
        )

        if matches:
            results.append({
                "job_id": row["job_id"],
                "job_title": row["job_title"],
                "company": row["company_name"],
                "matched_count": len(matches),
                "matched_keywords": ", ".join(matches),
                "description": row["cleaned_description"]
            })

    results_df = pd.DataFrame(results)

    if results_df.empty:
        print("No keyword matches found.")
        return results_df

    results_df = results_df.sort_values(
        by="matched_count",
        ascending=False
    ).reset_index(drop=True)

    print(f"Query: {query}")
    print(f"Jobs with at least one keyword match: {len(results_df)}")

    return results_df
```

```python
query = "What programming languages are required for machine learning engineering?"

keywords = [
    "programming languages",
    "Python",
    "C/C++",
    "C++",
    "Go",
    "Java",
    "JavaScript",
    "SQL",
    "Scala",
    "R"
]

suggestions = suggest_relevant_jobs(query, keywords)

display(
    suggestions[
        ["job_id", "job_title", "company",
         "matched_count", "matched_keywords"]
    ]
)
```
